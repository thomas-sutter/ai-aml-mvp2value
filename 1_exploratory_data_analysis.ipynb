# **EDA IBM Synthetic Data Sets for Core Banking and Money Laundering**

This notebook follows a practical workflow that combines the **PACE** framework (Plan, Analyze, Construct, Execute) with the six practices of the **Exploratory Data Analysis (EDA)** process.

#### **The 6 Practices of EDA**



*   Discovering
*   Struturing
*   Cleaning
*   Joining
*   Validating
*   Presenting




Exploratory Data Analysis (EDA) is not a fixed recipe or a linear, step-by-step procedure. The six EDA practices are **iterative** and **non-sequential**:

*   **Iterative**: a process repeated multiple times, with each pass building on prior results.
*   **Non-sequential**: not constrained to a predetermined order or sequence.

Because datasets vary, the exploration strategy will differ from case to case. Throughout EDA, logic and prior experience determine which of the six practices to apply, how many times to cycle through them, and when to apply them.

# Description

**IBM Synthetic Data Sets** for **Core Banking and Money Laundering** supports **AI model training** for essential banking services. This dataset simulates an entire banking ecosystem with lists of bank transfers, personal accounts for individuals, and corporate accounts for companies. It is specialized to find and label illegal banking transactions, such as check fraud, money laundering, and automated push payment (APP) fraud.

Because money laundering often goes undetected, having a dataset that is specialized in **identifying transactions** for **fraud** and **money laundering** is highly valuable. The dataset helps models determine the type of laundering, for example, fan-in, fan-out, or cycle. As a result, Synthetic Data Sets for Core Banking and Money Laundering can **offer key insights** for **creating** an **anti-money laundering solution**.

IBM Synthetic Data Sets for Core Banking and Money Laundering is best suited for the following **business use cases**:

*   **Money laundering detection**
*   **Check fraud**
*   **APP fraud**
*   **Loan default prediction**
*   **Honeypot**

## Calculate total transactions

### Subtask:
Calculate the total number of transactions across all banks from the `Num_Transactions` column of the `banks_df` DataFrame.


**Reasoning**:
Calculate the sum of the `Num_Transactions` column in the `banks_df` DataFrame and store it in `total_transactions`.



```python
total_transactions = banks_df['Num_Transactions'].sum()
print("Total number of transactions across all banks:", total_transactions)
```

## Sort banks by transactions

### Subtask:
Sort the `banks_df` DataFrame in descending order based on the `Num_Transactions` column.


**Reasoning**:
Sort the DataFrame by 'Num_Transactions' in descending order and display the head.



```python
banks_df = banks_df.sort_values(by='Num_Transactions', ascending=False)
display(banks_df.head())
```

## Calculate cumulative transactions and percentage

### Subtask:
Calculate the cumulative sum of `Num_Transactions` and the cumulative percentage of total transactions for each bank in the sorted `banks_df`.


**Reasoning**:
Calculate the cumulative sum and cumulative percentage of transactions and display the updated dataframe.



```python
banks_df['Cumulative_Transactions'] = banks_df['Num_Transactions'].cumsum()
banks_df['Cumulative_Transaction_Percentage'] = (banks_df['Cumulative_Transactions'] / total_transactions) * 100
display(banks_df.head())
```

```python
def parse_country(sample):
    s = str(sample)
    if "Online" in s:
        return "Online"
    parts = [p.strip() for p in s.split("-")]
    return parts[-1] if parts else np.nan

banks["Sample_Country"] = banks["Sample_City"].map(parse_country)
banks["Is_Online"] = banks["Num_Total_Locations"].eq(0)
banks["Is_International"] = banks["Num_non_Focus_Country_Locns"].gt(0)

print("Country distribution (Top 10):")
print(banks["Sample_Country"].value_counts().head(10))

# Namens-Duplikate (können Joins stören)
name_dups = banks["Bank_Name"].value_counts()
dups = name_dups[name_dups > 1]
print(f"\nBank_Name duplicates: {len(dups)} names duplicated, total duplicate rows: {len(banks) - banks['Bank_Name'].nunique()}")
display(dups.head(10))
```

```python
tx = banks["Num_Transactions"].astype(float)
shares = tx / tx.sum()

# Herfindahl-Hirschman-Index (HHI): Summe der quadrierten Marktanteile (in Prozent)
HHI = ((shares * 100.0)**2).sum()

def gini(arr):
    arr = np.asarray(arr, dtype=float)
    arr = arr[arr >= 0]
    if arr.size == 0:
        return np.nan
    arr.sort()
    n = arr.size
    cum = np.cumsum(arr)
    return (n + 1 - 2*np.sum(cum) / cum[-1]) / n

Gini_tx = gini(tx.values)

print(f"HHI (Transactions): {HHI:.0f}")
print(f"Gini (Transactions): {Gini_tx:.3f}")

# Top-1% und Top-10%
n = len(banks)
top1 = max(1, int(round(n * 0.01)))
top10 = int(round(n * 0.10))
top1_share = tx.nlargest(top1).sum() / tx.sum()
top10_share = tx.nlargest(top10).sum() / tx.sum()
print(f"Top 1% share: {top1_share:.3%} | Top 10% share: {top10_share:.3%}")
```

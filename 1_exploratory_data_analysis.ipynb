This notebook follows a practical workflow that combines the **PACE** framework (Plan, Analyze, Construct, Execute) with the six practices of the **Exploratory Data Analysis (EDA)** process.

#### **The 6 Practices of EDA**



*   Discovering
*   Struturing
*   Cleaning
*   Joining
*   Validating
*   Presenting




Exploratory Data Analysis (EDA) is not a fixed recipe or a linear, step-by-step procedure. The six EDA practices are **iterative** and **non-sequential**:

*   **Iterative**: a process repeated multiple times, with each pass building on prior results.
*   **Non-sequential**: not constrained to a predetermined order or sequence.

Because datasets vary, the exploration strategy will differ from case to case. Throughout EDA, logic and prior experience determine which of the six practices to apply, how many times to cycle through them, and when to apply them.

# Description

**IBM Synthetic Data Sets** for **Core Banking and Money Laundering** supports **AI model training** for essential banking services. This dataset simulates an entire banking ecosystem with lists of bank transfers, personal accounts for individuals, and corporate accounts for companies. It is specialized to find and label illegal banking transactions, such as check fraud, money laundering, and automated push payment (APP) fraud.

```python
# 0_SETUP — run first after opening
import pandas as pd, numpy as np
pd.options.display.float_format = '{:,.2f}'.format  # 2 Dezimalstellen in der Anzeige

def validate_balances(df, tol=1e-6):
    need = {"From_Initial_Balance","From_End_Balance","Amount_Paid",
            "To_Initial_Balance","To_End_Balance","Amount_Received"}
    if not need.issubset(df.columns):
        return pd.DataFrame([{"issue":"missing_columns","rows":0,"rate":0.0}])
    for c in need: df[c] = pd.to_numeric(df[c], errors='coerce')
    exp_from_end = df["From_Initial_Balance"] - df["Amount_Paid"]
    exp_to_end   = df["To_Initial_Balance"] + df["Amount_Received"]
    bad_from = (df["From_End_Balance"] - exp_from_end).abs() > tol
    bad_to   = (df["To_End_Balance"]   - exp_to_end).abs()   > tol
    return pd.DataFrame([
        {"issue":"from_balance_mismatch","rows":int(bad_from.sum()),"rate":float(bad_from.mean())},
        {"issue":"to_balance_mismatch",  "rows":int(bad_to.sum()),  "rate":float(bad_to.mean())},
    ])
```

## **Banks: sd175_Small_banks.csv**

### sd175_Small_banks.csv — Discovering (Impact → Insight)

**Quick profile**
- Shape: **1,458 rows × 6 columns** (`Bank_ID`, `Bank_Name`, `Num_Transactions`, `Num_Total_Locations`, `Num_non_Focus_Country_Locns`, `Sample_City`); **0 missing**.
- Keys: `Bank_ID` **unique** (1,458); `Bank_Name` **not unique** (1,330 unique; **128** duplicate rows across **62** names).
- Scale: **98.5%** of banks have **1** location; **5** online-only banks (0 locations).
- Volume: transactions are **highly skewed** (median **73.5**, max **130,718**); **corr**(`Num_Total_Locations`,`Num_Transactions`) ≈ **0.931**.

**Market concentration (why it matters)**
- **Top 1% (15 banks)** process **~72.6%** of all transactions; **Top 10% (146 banks)** process **~87.4%** → extreme Pareto pattern.
- Online-only: **5 banks ≈ 3.5%** of transactions. International footprint (`Num_non_Focus_Country_Locns`>0): **24 banks ≈ 25.1%** of transactions.

**Key takeaways**
- The table is a **lean bank dimension** (intentionally “short on data”): perfect for **joining** with accounts and transfers rather than standalone analysis.
- **Use IDs for joins** wherever possible; `Bank_Name` is ambiguous and must be normalized if used.
- Because of heavy skew and concentration, **rate-based metrics** (per 1,000 transactions/locations) are more informative than raw counts.

**Key insights (Impact → Insight)**
- **Few giants, long tail:** a handful of multi-location banks dominate volume; the many one-branch banks form a clean “tail” for anomaly signals (fan-in/fan-out, sudden spikes).
- **Segment risk by footprint:** treat **online-only** and **cross-border** banks as distinct risk strata; patterns and baselines differ materially.
- **Scaling effects:** strong size–volume coupling (r≈0.93) suggests normalizing by footprint to avoid flagging “big because big.”

**Where it fits in the Core Banking & Money Laundering dataset**
- Acts as the **institution dimension** to enrich:
  - `sd175_Small_liquid_accts_people.csv` / `…_companies.csv` → customers/accounts per bank.
  - `sd175_Small_bank_xfers.csv` → payment flows (from/to bank), cross-border topology.
  - `sd175_Small_b2b.csv` → corporate relationships.
- Enables features like: **Tx per branch**, **cross-border Tx rate**, **online-only Tx share**, **concentration indices (HHI/Gini)** per segment.

**EDA process hooks**
- **Discovering:** profile schema, uniqueness, missingness, skew, concentration (this cell).
- **Validating:** check key integrity for joins; confirm name normalization if needed.
- **Structuring/Joining:** merge with accounts/transfers; derive footprint- and segment-normalized features.
- **Explaining:** log-scale visuals (scatter locations vs. transactions), Top-N bars, Lorenz curves to communicate concentration.

**PACE alignment**
- **Plan:** questions on bank footprint vs. volume, online/cross-border risk strata.
- **Analyze:** quantify skew, concentration, correlations; identify join risks (`Bank_Name`).
- **Construct:** build clean joins via `Bank_ID`, create normalized features and segments.
- **Execute:** produce focused visuals & rates; feed features to AML/Fraud models and monitoring dashboards.

```python
def parse_country(sample):
    s = str(sample)
    if "Online" in s:
        return "Online"
    parts = [p.strip() for p in s.split("-")]
    return parts[-1] if parts else np.nan

banks_df["Sample_Country"] = banks_df["Sample_City"].map(parse_country)
banks_df["Is_Online"] = banks_df["Num_Total_Locations"].eq(0)
banks_df["Is_International"] = banks_df["Num_non_Focus_Country_Locns"].gt(0)

print("Country distribution (Top 10):")
print(banks_df["Sample_Country"].value_counts().head(10))

# Namens-Duplikate (können Joins stören)
name_dups = banks_df["Bank_Name"].value_counts()
dups = name_dups[name_dups > 1]
print(f"\nBank_Name duplicates: {len(dups)} names duplicated, total duplicate rows: {len(banks_df) - banks_df['Bank_Name'].nunique()}")
display(dups.head(10))


tx = banks_df["Num_Transactions"].astype(float)
shares = tx / tx.sum()

# Herfindahl-Hirschman-Index (HHI): Summe der quadrierten Marktanteile (in Prozent)
HHI = ((shares * 100.0)**2).sum()

def gini(arr):
    arr = np.asarray(arr, dtype=float)
    arr = arr[arr >= 0]
    if arr.size == 0:
        return np.nan
    arr.sort()
    n = arr.size
    cum = np.cumsum(arr)
    return (n + 1 - 2*np.sum(cum) / cum[-1]) / n

Gini_tx = gini(tx.values)

print(f"HHI (Transactions): {HHI:.0f}")
print(f"Gini (Transactions): {Gini_tx:.3f}")

# Top-1% und Top-10%
n = len(banks_df)
top1 = max(1, int(round(n * 0.01)))
top10 = int(round(n * 0.10))
top1_share = tx.nlargest(top1).sum() / tx.sum()
top10_share = tx.nlargest(top10).sum() / tx.sum()
print(f"Top 1% share: {top1_share:.3%} | Top 10% share: {top10_share:.3%}")
```

#### FK-Coverage Check

```python
import re, numpy as np, pandas as pd

banks  = banks_df.copy()
people = liquid_accts_people_df.copy()

# Sanity checks
need_banks  = {"Bank_ID","Bank_Name"}
need_people = {"Financial_Institution_ID","Financial_Institution_Name"}
missing_b = need_banks  - set(banks.columns)
missing_p = need_people - set(people.columns)
assert not missing_b, f"banks_df missing columns: {missing_b}"
assert not missing_p, f"liquid_accts_people_df missing columns: {missing_p}"

# Normalizer für Namen
def norm(s):
    if pd.isna(s): return ""
    s = str(s).lower().strip()
    s = re.sub(r"[\.\,\-_/&()]+", " ", s)
    s = re.sub(r"\s+", " ", s)
    return s

# Coverage by ID
left_id  = people["Financial_Institution_ID"].astype(str).str.strip()
right_id = banks["Bank_ID"].astype(str).str.strip()
cov_id_mask = left_id.isin(right_id)
cov_id = cov_id_mask.mean()

# Coverage by Name (normalized)
left_nm  = people["Financial_Institution_Name"].map(norm)
right_nm = banks["Bank_Name"].map(norm)
cov_nm_mask = left_nm.isin(right_nm)
cov_nm = cov_nm_mask.mean()

# „Cash“ ausnehmen (typischer Sonderfall, steht meist nicht in Banks)
is_cash = left_nm.eq("cash")
cov_id_wo_cash = cov_id_mask[~is_cash].mean()
cov_nm_wo_cash = cov_nm_mask[~is_cash].mean()

print(f"Coverage by ID: {cov_id:.1%} (w/o 'cash': {cov_id_wo_cash:.1%})")
print(f"Coverage by Name (normalized): {cov_nm:.1%} (w/o 'cash': {cov_nm_wo_cash:.1%})")

# Diagnose: nicht gematchte Institute
unmatched = people.loc[~(cov_id_mask | cov_nm_mask), "Financial_Institution_Name"]
print("Unmatched rows (incl. cash):", int(unmatched.shape[0]))
display(unmatched.value_counts().head(15))
```

```python
# Fallback-Mapping über normalisierte Banknamen -> Bank_ID
bnorm_map = (
    banks.assign(bank_name_norm=banks["Bank_Name"].map(norm))
         .drop_duplicates("bank_name_norm")
         .set_index("bank_name_norm")["Bank_ID"]
         .to_dict()
)

people = people.copy()
people["bank_id_by_id"]   = left_id.where(left_id.isin(right_id), np.nan)
people["bank_id_by_name"] = people["Financial_Institution_Name"].map(norm).map(bnorm_map)

# Prefer ID, then fallback to name
people["Resolved_Bank_ID"] = people["bank_id_by_id"].fillna(people["bank_id_by_name"]).astype("string")

resolved_cov = people["Resolved_Bank_ID"].notna().mean()
resolved_cov_wo_cash = people.loc[~is_cash, "Resolved_Bank_ID"].notna().mean()
print(f"Resolved coverage (ID or Name): {resolved_cov:.1%} (w/o 'cash': {resolved_cov_wo_cash:.1%})")

# Optional: Bank-Attribute anhängen (falls Spalten existieren)
keep_cols = ["Bank_ID","Bank_Name"]
for c in ["Num_Total_Locations","Num_non_Focus_Country_Locns","Sample_City","Is_Online","Is_CrossBorder"]:
    if c in banks.columns: keep_cols.append(c)

banks_small = banks[keep_cols].rename(columns={"Bank_ID":"Resolved_Bank_ID"})
people_enriched = people.merge(banks_small, on="Resolved_Bank_ID", how="left")

# Sonderfall: stabiler Account-Key für 'Cash'-Zeilen (Account_ID==0)
if "Account_ID" in people_enriched.columns and "Entity_ID" in people_enriched.columns and "Account_Type" in people_enriched.columns:
    people_enriched["Account_Key_safe"] = np.where(
        people_enriched["Account_ID"].eq(0),
        people_enriched["Entity_ID"].astype(str) + "|" + people_enriched["Account_Type"].astype(str) + "|" + people_enriched["Financial_Institution_Name"].astype(str),
        people_enriched["Account_ID"].astype(str)
    )

display(people_enriched.head(3))
```

# 2_Joining — Create xfers_enriched with account, bank, and B2B context

```python
import pandas as pd, numpy as np, re

BASE_PATH = '/content/drive/MyDrive/Portfolio/AML/core_banking_and_money_laundering/data'
PATHS = dict(
    banks     = f'{BASE_PATH}/sd175_Small_banks.csv',
    people    = f'{BASE_PATH}/sd175_Small_liquid_accts_people.csv',
    companies = f'{BASE_PATH}/sd175_Small_liquid_accts_companies.csv',
    xfers     = f'{BASE_PATH}/sd175_Small_bank_xfers.csv',
    out       = f'{BASE_PATH}/xfers_enriched_v2.csv',
)

def read_csv(path, **kw):
    return pd.read_csv(path, low_memory=False, **kw)

def to_key(s: pd.Series) -> pd.Series:
    """Normalize join keys to string; strip, drop trailing .0, keep NaNs."""
    out = s.astype(str)
    # turn literal 'nan'/'<NA>' back into NaN after astype(str)
    out = out.where(~out.str.lower().isin(['nan','<na>','none','null']), other=np.nan)
    return (out.str.replace(r'\.0$', '', regex=True).str.strip())

def build_acct_dim(people, companies):
    # nur vorhandene Spalten nehmen (IBM-CSV sind stabil, aber wir gehen defensiv vor)
    want = ['Account_ID','Entity_Type','Entity_ID','Entity_Name',
            'Account_Country','Account_Currency','MCC','Account_Type',
            'Financial_Institution_ID','Financial_Institution_Name']
    keep_p = [c for c in want if c in people.columns]
    keep_c = [c for c in want if c in companies.columns]
    p = people[keep_p].copy()
    c = companies[keep_c].copy()

    # Keys & IDs als Strings
    if 'Account_ID' in p: p['Account_ID'] = to_key(p['Account_ID'])
    if 'Account_ID' in c: c['Account_ID'] = to_key(c['Account_ID'])
    for df in (p,c):
        if 'Financial_Institution_ID' in df:
            df['Financial_Institution_ID'] = to_key(df['Financial_Institution_ID'])

    dim = (pd.concat([p,c], ignore_index=True)
             .drop_duplicates(subset=['Account_ID']) if 'Account_ID' in (set(p.columns)|set(c.columns)) else pd.concat([p,c], ignore_index=True))
    return dim

def derive_features(df):
    num = ['Amount_Paid','Amount_Received','From_Initial_Balance','From_End_Balance',
           'To_Initial_Balance','To_End_Balance']
    for c in num:
        if c in df.columns: df[c] = pd.to_numeric(df[c], errors='coerce')

    if {'From_Bank_ID','To_Bank_ID'}.issubset(df.columns):
        df['Is_Cross_Bank'] = (df['From_Bank_ID'] != df['To_Bank_ID']).astype('Int64')

    if {'Payment_Currency','Receiving_Currency'}.issubset(df.columns):
        df['Is_FX'] = (df['Payment_Currency'] != df['Receiving_Currency']).astype('Int64')

    ts_col = next((c for c in ['Timestamp','Txn_Timestamp','Datetime','Date'] if c in df.columns), None)
    if ts_col:
        ts = pd.to_datetime(df[ts_col], errors='coerce')
        df['Txn_TS']    = ts
        df['Txn_Year']  = ts.dt.year
        df['Txn_Month'] = ts.dt.month
        df['Txn_DOW']   = ts.dt.dayofweek
        df['Txn_Hour']  = ts.dt.hour
        df['Is_Weekend']= ts.dt.dayofweek.isin([5,6]).astype('Int64')

    need = {'From_Initial_Balance','From_End_Balance','Amount_Paid',
            'To_Initial_Balance','To_End_Balance','Amount_Received'}
    if need.issubset(df.columns):
        tol = 1e-6
        df['From_Balance_Mismatch'] = ((df['From_End_Balance'] - (df['From_Initial_Balance'] - df['Amount_Paid'])).abs() > tol).astype('Int64')
        df['To_Balance_Mismatch']   = ((df['To_End_Balance']   - (df['To_Initial_Balance'] + df['Amount_Received'])).abs() > tol).astype('Int64')
    return df

def build_xfers_enriched_v2(PATHS):
    banks     = read_csv(PATHS['banks'], dtype=str)
    people    = read_csv(PATHS['people'], dtype=str)
    try:
        companies = read_csv(PATHS['companies'], dtype=str)
    except UnicodeDecodeError:
        companies = read_csv(PATHS['companies'], dtype=str, encoding='latin1')
    # WICHTIG: xfers auch als STRING laden, damit Join-Keys kompatibel sind
    xfers     = read_csv(PATHS['xfers'], dtype=str)

    # Join-Keys in xfers normalisieren
    for col in ['From_Account','To_Account']:
        if col in xfers.columns:
            xfers[col] = to_key(xfers[col])

    # Account-Dimension bauen & Keys normalisieren
    acct_dim = build_acct_dim(people, companies)
    if 'Account_ID' in acct_dim.columns:
        acct_dim['Account_ID'] = to_key(acct_dim['Account_ID'])

    # Join via Accounts (From/To)
    if 'From_Account' in xfers.columns and 'Account_ID' in acct_dim.columns:
        X = xfers.merge(acct_dim.add_prefix('From_'),
                        left_on='From_Account', right_on='From_Account_ID', how='left')
    else:
        raise KeyError("Erwarte Spalten 'From_Account' in xfers und 'Account_ID' in Accounts.")

    if 'To_Account' in xfers.columns and 'Account_ID' in acct_dim.columns:
        X = X.merge(acct_dim.add_prefix('To_'),
                    left_on='To_Account', right_on='To_Account_ID', how='left')
    else:
        raise KeyError("Erwarte Spalten 'To_Account' in xfers und 'Account_ID' in Accounts.")

    # Kanonische Spalten
    X['From_Party_Type'] = X.get('From_Entity_Type')
    X['From_Party_ID']   = X.get('From_Entity_ID')
    X['From_Party_Name'] = X.get('From_Entity_Name')
    X['From_Bank_ID']    = X.get('From_Financial_Institution_ID')
    X['From_Bank_Name']  = X.get('From_Financial_Institution_Name')

    X['To_Party_Type'] = X.get('To_Entity_Type')
    X['To_Party_ID']   = X.get('To_Entity_ID')
    X['To_Party_Name'] = X.get('To_Entity_Name')
    X['To_Bank_ID']    = X.get('To_Financial_Institution_ID')
    X['To_Bank_Name']  = X.get('To_Financial_Institution_Name')

    # Optionale Bank-Attribute (falls vorhanden)
    bank_cols = [c for c in ['Bank_ID','Bank_Name','Num_Transactions','Num_Total_Locations',
                             'Num_non_Focus_Country_Locns','Sample_City'] if c in banks.columns]
    if 'Bank_ID' in banks.columns:
        bank_attr = banks[bank_cols].drop_duplicates('Bank_ID')
        bf = bank_attr.rename(columns={
            'Bank_ID':'From_Bank_ID','Bank_Name':'From_Bank_Name_Master',
            'Num_Transactions':'From_Bank_Num_Transactions',
            'Num_Total_Locations':'From_Bank_Num_Total_Locations',
            'Num_non_Focus_Country_Locns':'From_Bank_Num_nonFocus_Locns',
            'Sample_City':'From_Bank_Sample_City'
        })
        bt = bank_attr.rename(columns={
            'Bank_ID':'To_Bank_ID','Bank_Name':'To_Bank_Name_Master',
            'Num_Transactions':'To_Bank_Num_Transactions',
            'Num_Total_Locations':'To_Bank_Num_Total_Locations',
            'Num_non_Focus_Country_Locns':'To_Bank_Num_nonFocus_Locns',
            'Sample_City':'To_Bank_Sample_City'
        })
        X = X.merge(bf, on='From_Bank_ID', how='left')
        X = X.merge(bt, on='To_Bank_ID',   how='left')

    # Features & Coverage
    X = derive_features(X)
    cov = X[['From_Bank_ID','To_Bank_ID','From_Party_Type','To_Party_Type']].isna().mean().sort_values(ascending=False)
    print('Missing rate after account-based mapping:\n', cov)

    X.to_csv(PATHS['out'], index=False)
    print('Saved:', PATHS['out'], '| rows:', len(X), '| cols:', len(X.columns))
    return X

_ = build_xfers_enriched_v2(PATHS)
```

### What `Is_Self_Transfer` means

`Is_Self_Transfer = 1` marks transfers where **sender and receiver are the same party** (same `Party_ID`).

**Why this is useful:**

* **Normal behavior:** People often move money between **their own accounts** (e.g., checking → savings, personal → credit card). That’s fine.
* **Risk signal in context:** If this happens **very often**, **in short time windows**, with **unusual amounts**, or **across banks/countries**, it can indicate **layering/obfuscation**.
* **Strongest when combined with other features:**

  * `Is_Self_Transfer = 1` **and** `Is_Cross_Bank = 1` → moving own funds **between two banks**.
  * **Bursts** of self-transfers (e.g., ≥5 within 24h) → automation/pattern.
  * **Amounts just below thresholds** (e.g., ~9,900) → structuring.
  * Self-transfer **plus FX change** → added obfuscation signal.
* **Interpretation:** On its own it’s **not evidence**, just a **weak signal**. In ML, it helps **together with time-, amount-, and network-based features** to separate benign from risky behavior.

**Bottom line:** `Is_Self_Transfer` is a **context flag**: it says the money stayed with the **same beneficial owner**. Your model, using additional signals, decides whether it’s **harmless** (saving, credit card payments) or **suspicious** (layering, round-tripping).

```python
# 3_VALIDATING (robust v2) — safe to paste & run

import pandas as pd, numpy as np

BASE_PATH = '/content/drive/MyDrive/Portfolio/AML/core_banking_and_money_laundering/data'
PATHS = dict(
    xfers_raw = f'{BASE_PATH}/sd175_Small_bank_xfers.csv',
    enr_v3    = f'{BASE_PATH}/xfers_enriched_v3.csv',
)

# Load enriched with safe dtypes (avoid category for currencies in validation)
preview_cols = pd.read_csv(PATHS['enr_v3'], nrows=1).columns
parse_dates = [c for c in ['Txn_TS','Timestamp','Txn_Timestamp','Datetime','Date'] if c in preview_cols]

dtype_map = {
    'From_Bank_ID':'string','To_Bank_ID':'string',
    'From_Party_ID':'string','To_Party_ID':'string',
    'From_Party_Type':'category','To_Party_Type':'category',
    # keep currencies as string here to avoid categorical-compare errors
    'Payment_Currency':'string','Receiving_Currency':'string',
}
enr = pd.read_csv(PATHS['enr_v3'], dtype=dtype_map, parse_dates=parse_dates, low_memory=False)

# --- 3.1 Key coverage
keys = ['From_Bank_ID','To_Bank_ID','From_Party_Type','To_Party_Type']
print('Key missing rates:\n', enr[keys].isna().mean().sort_values(ascending=False).round(6))

# --- 3.2 Duplicate detection: Txn-ID if present, else composite surrogate
id_candidates = [c for c in ['Transaction_ID','Txn_ID','ID'] if c in enr.columns]
if id_candidates:
    tx_id_col = id_candidates[0]
    dup_rate = float(enr.duplicated(subset=[tx_id_col]).mean())
    print('Duplicate rate on', tx_id_col, ':', round(dup_rate, 6))
else:
    composite = [c for c in ['Txn_TS','Timestamp','From_Account','To_Account','Amount_Paid','Amount_Received'] if c in enr.columns]
    dup_rate = float(enr.duplicated(subset=composite).mean()) if composite else np.nan
    print('Duplicate rate (composite key):', round(dup_rate, 6), '| subset:', composite)

# --- 3.3 Flag consistency (cast to string before comparisons)
checks = {}
if {'From_Bank_ID','To_Bank_ID'}.issubset(enr.columns):
    cross_calc = (enr['From_Bank_ID'].astype('string') != enr['To_Bank_ID'].astype('string')).astype('Int64')
    if 'Is_Cross_Bank' in enr.columns:
        checks['cross_bank_consistency'] = float((enr['Is_Cross_Bank'].astype('Int64') == cross_calc).mean())

if {'From_Party_ID','To_Party_ID'}.issubset(enr.columns):
    self_calc = (enr['From_Party_ID'].astype('string') == enr['To_Party_ID'].astype('string')).astype('Int64')
    if 'Is_Self_Transfer' in enr.columns:
        checks['self_transfer_consistency'] = float((enr['Is_Self_Transfer'].astype('Int64') == self_calc).mean())

if {'Payment_Currency','Receiving_Currency'}.issubset(enr.columns):
    fx_calc = (enr['Payment_Currency'].astype('string') != enr['Receiving_Currency'].astype('string')).astype('Int64')
    if 'Is_FX' in enr.columns:
        checks['fx_flag_consistency'] = float((enr['Is_FX'].astype('Int64') == fx_calc).mean())
    else:
        checks['fx_flag_consistency'] = np.nan  # not present in file; just reported

print('Flag consistency:', {k: round(v, 6) if pd.notna(v) else v for k,v in checks.items()})

# --- 3.4 Amounts & balances
for c in ['Amount_Paid','Amount_Received','From_Initial_Balance','From_End_Balance','To_Initial_Balance','To_End_Balance']:
    if c in enr.columns:
        enr[c] = pd.to_numeric(enr[c], errors='coerce')

neg = {c: int((enr[c] < 0).sum()) for c in ['Amount_Paid','Amount_Received'] if c in enr.columns}
zero= {c: int((enr[c] == 0).sum()) for c in ['Amount_Paid','Amount_Received'] if c in enr.columns}
print('Negative counts:', neg, '| Zero counts:', zero)

def validate_balances(df, tol=1e-6):
    need = {"From_Initial_Balance","From_End_Balance","Amount_Paid",
            "To_Initial_Balance","To_End_Balance","Amount_Received"}
    if not need.issubset(df.columns):
        return pd.DataFrame([{"issue":"missing_columns","rows":0,"rate":0.0}])
    exp_from_end = df["From_Initial_Balance"] - df["Amount_Paid"]
    exp_to_end   = df["To_Initial_Balance"] + df["Amount_Received"]
    bad_from = (df["From_End_Balance"] - exp_from_end).abs() > tol
    bad_to   = (df["To_End_Balance"]   - exp_to_end).abs()   > tol
    return pd.DataFrame([
        {"issue":"from_balance_mismatch","rows":int(bad_from.sum()),"rate":float(bad_from.mean())},
        {"issue":"to_balance_mismatch",  "rows":int(bad_to.sum()),  "rate":float(bad_to.mean())}
    ])

print('Balance validation:\n', validate_balances(enr))

# --- 3.5 Time coverage
ts_col = next((c for c in ['Txn_TS','Timestamp','Txn_Timestamp','Datetime','Date'] if c in enr.columns), None)
if ts_col:
    ts = pd.to_datetime(enr[ts_col], errors='coerce')
    print('Timestamp coverage:', {'parse_fail_rate': float(ts.isna().mean()),
                                  'min': str(ts.min()), 'max': str(ts.max())})

# --- 3.6 Label sanity & imbalance (if any label present)
LABEL_COL = next((c for c in ['Is_Money_Laundering','Is_Fraud','Label_Is_ML','Is_APP_Fraud','Is_Check_Fraud'] if c in enr.columns), None)
if LABEL_COL:
    y = pd.to_numeric(enr[LABEL_COL], errors='coerce')
    print('Label coverage:', {'missing_rate': float(y.isna().mean()),
                              'positives_rate': float((y==1).mean())})
    # quick smoke correlations (not for inference, just to flag obvious leakage)
    quick_corrs = {}
    for c in ['Is_Self_Transfer','Is_Cross_Bank','Is_FX']:
        if c in enr.columns:
            quick_corrs[c] = float(pd.Series(enr[c]).corr(y))
    print('Quick correlations (leakage smoke-test):', quick_corrs)

# --- 3.7 Row count parity (raw vs enriched)
raw_rows = pd.read_csv(PATHS['xfers_raw'], low_memory=False).shape[0]
print({'raw_xfers': raw_rows, 'enriched_v3': len(enr), 'delta': len(enr)-raw_rows})
```
